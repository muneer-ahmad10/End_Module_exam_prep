# **Object Detection & Segmentation, YOLO**

### **Classification vs Detection vs Segmentation — the key distinction**

* **Classification:** "what is in this image?" → one label for the whole image

* **Object Detection:** "what objects are in this image, and where?" → multiple labels + bounding boxes (x, y, width, height for each object)

* **Segmentation:** "which exact pixels belong to which object?" → pixel-level classification, not just a box

**Two types of segmentation:**
* **Semantic segmentation:** labels every pixel by class (e.g. all "car" pixels get the same label), but doesn't distinguish between two different cars

* **Instance segmentation:** labels every pixel by class and which specific object instance it belongs to (Car #1 vs Car #2 get different labels even though both are "car")

### **Bounding boxes and IoU (Intersection over Union)**

IoU measures how well a predicted box overlaps with the ground-truth box:

***IoU = (Area of Overlap) / (Area of Union)***
* IoU = 1 → perfect overlap
* IoU = 0 → no overlap at all
* Commonly, a prediction is considered "correct" if IoU > 0.5 (a common threshold, though it can vary by benchmark)

NAT-style practice: if predicted box overlap area = 40, and union area (predicted + ground truth − overlap) = 100, then IoU = 40/100 = 0.4.

## **The Core Idea: One Single Forward Pass**

Older object detectors (like Faster R-CNN) work in two stages:

* First, scan the image to propose region candidates where objects might be.

* Second, run a CNN over each proposed region to classify what is inside.

YOLO does both simultaneously in a single pass. It looks at the full image once, splits it into a grid, and directly outputs bounding box coordinates and class probabilities all at the same time.

## **How YOLO Works Step-by-Step**

* **Step 1: Divide the Image into a Grid**
YOLO overlays an $S \times S$ grid (e.g., $7 \times 7$ or $13 \times 13$) over the input image.
***Rule of Responsibility:*** Whichever grid cell contains the center point of an object is responsible for detecting that object.

* **Step 2: Every Grid Cell Predicts a Output Vector**
Each grid cell predicts a vector containing three key pieces of information:

  Output Vector per Cell = [ p_c, b_x, b_y, b_h, b_w, c_1, c_2, ..., c_C ]

    1.   $p_c$ (Objectness Score): Is there an object centered in this cell? ($0$ or $1$)
        

          *   $b_x, b_y$: Center position of the bounding box (relative to the grid cell).
          *   $b_h, b_w$: Height and Width of the bounding box (relative to the full image).


    2.   Bounding Box Coordinates ($b_x, b_y, b_h, b_w$):
    3.   Class Probabilities ($c_1, c_2, \dots$): If an object exists, which class is it? (e.g., Car, Person, Dog).



* **Step 3: Non-Maximum Suppression (NMS)**
Because multiple nearby grid cells might predict overlapping boxes for the exact same object, YOLO cleans up duplicate predictions using Non-Maximum Suppression (NMS):



  1.   Discard all bounding boxes where objectness score $p_c < \text{threshold}$ (e.g., $0.6$).
  2.  Pick the box with the highest remaining $p_c$ score.
  3.  Calculate Intersection over Union (IoU) with all other overlapping boxes.
  4.  Remove any overlapping box where $\text{IoU} > 0.5$ (since it's a duplicate prediction for the same object).



## **Anchor Boxes**

if two objects have their centers in the exact same grid cell (e.g., a person standing in front of a car)?


*   Solution: Assign predefined box shapes called Anchor Boxes (e.g., one tall/skinny box for humans, one wide box for cars) to each grid cell so the cell can output multiple predictions simultaneously.

# **Autoencoders & GANs**

## **Autoencoders — the core idea**

An autoencoder learns to compress data down, then reconstruct it back to the original — trained with no labels (unsupervised), using the input itself as the target.



---


***Input → Encoder → Latent Space (bottleneck) → Decoder → Reconstructed Output***

---



**Loss function:** typically MSELoss between the input and the reconstructed output — the network is penalized based on how different the reconstruction is from the original.

### **Why the bottleneck matters:**  
by forcing data through a smaller-dimensional latent space than the original input, the network can't just memorize/copy — it's forced to learn only the most essential compressed features needed to reconstruct the input. This is the core reason autoencoders are useful for:

* **Dimensionality reduction** (the latent space itself is a compressed representation)
* **Denoising** (train it to reconstruct clean data from noisy input — it learns to ignore noise)
* **Anomaly detection** (data that doesn't reconstruct well is likely out-of-distribution/anomalous, since the network only learned to compress "normal" patterns well)

**Important exam distinction:** vanilla autoencoders are not reliable generative models — they reconstruct what you feed them, but sampling random points in latent space and decoding them usually produces garbage, because the latent space isn't structured to be smoothly sampleable. That's specifically what a Variational Autoencoder (VAE) fixes (by forcing the latent space into a smooth, structured distribution) — and it's a different thing from a GAN.

# **GANs (Generative Adversarial Networks)**

Two networks trained in direct competition:

Network	---------------|------------Job


---


**Generator**	 -------------: Takes random noise as input, tries to produce fake data realistic enough to fool the discriminator

**Discriminator** ---------:	Tries to distinguish real training data from the generator's fakes

**Training dynamic:** it's a minimax game — the generator improves by learning what fools the discriminator; the discriminator improves by learning to catch better fakes. Ideally, this adversarial pressure pushes the generator toward producing highly realistic synthetic data over many training iterations.

**Contrast with autoencoders (a common exam angle):** an autoencoder is optimized against a fixed target (its own input, via reconstruction loss) — a single, stable objective. A GAN's generator is optimized against a moving target (the discriminator, which is itself being trained simultaneously) — this adversarial, two-network setup is fundamentally different from an autoencoder's single-network reconstruction setup, and is why GANs are notoriously harder to train stably.
