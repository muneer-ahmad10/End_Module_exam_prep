In NumPy and PyTorch, passing -1 inside .reshape() tells the library to automatically calculate that dimension for you based on the total number of elements in the array.How It WorksAn array’s total number of elements must remain constant before and after reshaping:$$\text{Total Elements} = \text{Rows} \times \text{Columns}$$When you specify one dimension (e.g., 5 rows) and set the other dimension to -1, Python calculates:$$\text{Automatic Dimension (-1)} = \frac{\text{Total Elements}}{5}$$

# **Object Detection & Segmentation, YOLO**

### **Classification vs Detection vs Segmentation — the key distinction**

* **Classification:** "what is in this image?" → one label for the whole image

* **Object Detection:** "what objects are in this image, and where?" → multiple labels + bounding boxes (x, y, width, height for each object)

* **Segmentation:** "which exact pixels belong to which object?" → pixel-level classification, not just a box

**Two types of segmentation:**
* **Semantic segmentation:** labels every pixel by class (e.g. all "car" pixels get the same label), but doesn't distinguish between two different cars

* **Instance segmentation:** labels every pixel by class and which specific object instance it belongs to (Car #1 vs Car #2 get different labels even though both are "car")
